```python
import pandas as pd

bookings = pd.read_csv("goa_manali_bookings.csv")

print("Rows, Columns:", bookings.shape)

bookings.head()
```

**Converted Date Column**

```python
bookings['checkin_date'] = pd.to_datetime(bookings['checkin_date'])

bookings.info()
```

Checking how many booking of two different location has been done on different days

```python
goa_dates = bookings[bookings['property_city'] == 'Goa']['checkin_date'].dt.date.unique()

manali_dates = bookings[bookings['property_city'] == 'Manali']['checkin_date'].dt.date.unique()

print("Goa dates:", len(goa_dates))
print("Manali dates:", len(manali_dates))
```

API Function

```python
import requests
import pandas as pd

def get_weather(city, latitude, longitude, start_date, end_date):

    url = (
        f"https://archive-api.open-meteo.com/v1/era5?"
        f"latitude={latitude}&longitude={longitude}"
        f"&start_date={start_date}"
        f"&end_date={end_date}"
        f"&daily=precipitation_sum,temperature_2m_max"
    )

    try:
        response = requests.get(url, timeout=30)

        if response.status_code != 200:
            print(f"API Error for {city}")
            return pd.DataFrame()

        data = response.json()

        weather = pd.DataFrame({
            "date": data["daily"]["time"],
            "precipitation_sum": data["daily"]["precipitation_sum"],
            "temperature_max": data["daily"]["temperature_2m_max"]
        })

        weather["city"] = city

        return weather

    except Exception as e:
        print("Error:", e)
        return pd.DataFrame()
```

Fetch weather


```python
goa_weather = get_weather(
    "Goa",
    15.2993,
    74.1240,
    "2024-01-01",
    "2024-12-31"
)

manali_weather = get_weather(
    "Manali",
    32.2396,
    77.1887,
    "2024-01-01",
    "2024-12-31"
)

weather = pd.concat([goa_weather, manali_weather])

print(weather.shape)

weather.head()
```

prepare join keys


```python
weather["date"] = pd.to_datetime(weather["date"])

bookings["date"] = bookings["checkin_date"]

bookings.head()
```

Merge weather and Bookings

```python
merged = bookings.merge(
    weather,
    left_on=["property_city", "date"],
    right_on=["city", "date"],
    how="left"
)

print(merged.shape)

merged.head()
```

Create rain cateogry

```python
merged["rain_category"] = merged["precipitation_sum"].apply(
    lambda x: "Heavy Rain" if x >= 20 else "Normal Weather"
)

merged["rain_category"].value_counts()
```

Calculate cancellations

```python
merged["cancel_or_noshow"] = merged["booking_status"].isin(
    ["Cancelled", "No-Show"]
)

rain_analysis = merged.groupby("rain_category").agg(
    total_bookings=("booking_id", "count"),
    cancellations=("cancel_or_noshow", "sum")
)

rain_analysis["rate_pct"] = round(
    rain_analysis["cancellations"] * 100 /
    rain_analysis["total_bookings"],
    2
)

rain_analysis
```
